```python
def extract_chr(rsid):
    match = re.match(r'chr([0-9XY]+)_', rsid)
    if match:
        return match.group(1)
    return None
```

```python
gtex['Chromosome'] = gtex['RSID'].apply(extract_chr)
```

```python
chromosome_dfs = {chr_num: df_sub for chr_num, df_sub in gtex.groupby('Chromosome')}
```

```python
for chr_num, df_sub in chromosome_dfs.items():
    filename = f'/Users/s2096435/Library/CloudStorage/OneDrive-UniversityofEdinburgh/test_code_review/tarviz/chrom/chr{chr_num}.hdf5'
    df_sub.to_hdf(filename, key='RSID', mode='w')
```
